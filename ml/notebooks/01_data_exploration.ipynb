# 01 · Data exploration
UCI-HAR raw inertial signals: shapes, class balance, subjects, example windows.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))  # run from ml/notebooks
import numpy as np
import matplotlib.pyplot as plt
from src import config
from src.data_loader import load_split, load_subjects, activity_name
```

```python
X_train, y_train, _ = load_split('train')
X_test, y_test, _ = load_split('test')
print(X_train.shape, X_test.shape)
```

```python
for split, y in [('train', y_train), ('test', y_test)]:
    counts = np.bincount(y, minlength=6)
    print(split, {activity_name(i): int(c) for i, c in enumerate(counts)})
```

```python
print('train subjects', sorted(set(load_subjects('train'))))
print('test subjects', sorted(set(load_subjects('test'))))
```

```python
fig, axes = plt.subplots(6, 1, figsize=(10, 14), sharex=True)
for c, ax in enumerate(axes):
    i = int(np.where(y_train == c)[0][0])
    for ch in range(6, 9):
        ax.plot(X_train[i, :, ch], label=config.SIGNALS[ch])
    ax.set_title(activity_name(c)); ax.legend(loc='upper right')
plt.tight_layout()
```
